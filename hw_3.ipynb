```python
import pandas as pd

df = pd.read_csv('/content/train.csv')

print("=== Информация о датасете ===")
print(df.info())
print("\n=== Число пропусков по столбцам ===")
print(df.isnull().sum())
print("\n=== Основная статистика ===")
print(df.describe(include='all'))

print("\n=== Процент выживаемости по классам ===")
survival_rate = df.groupby('Pclass')['Survived'].mean() * 100
print(survival_rate)

def extract_first_name(name):
    name = name.split('(')[0]
    name = name.split(',')[1].strip()
    parts = name.split()
    return parts[-1] if len(parts) > 1 else parts[0]

df['FirstName'] = df['Name'].apply(extract_first_name)

male_names = df[df['Sex'] == 'male']['FirstName']
female_names = df[df['Sex'] == 'female']['FirstName']

print("\n=== Самое популярное мужское имя ===")
print(male_names.value_counts().head(1))

print("\n=== Самое популярное женское имя ===")
print(female_names.value_counts().head(1))

print("\n=== Самые популярные имена в каждом классе ===")
for cls in sorted(df['Pclass'].unique()):
    males = df[(df['Pclass'] == cls) & (df['Sex'] == 'male')]['FirstName']
    females = df[(df['Pclass'] == cls) & (df['Sex'] == 'female')]['FirstName']
    print(f"\nКласс {cls}:")
    if not males.empty:
        print("  Мужчины:", males.value_counts().head(1).to_dict())
    if not females.empty:
        print("  Женщины:", females.value_counts().head(1).to_dict())

print("\n=== Пассажиры старше 44 лет ===")
print(df[df['Age'] > 44].head())

print("\n=== Мужчины младше 44 лет ===")
print(df[(df['Age'] < 44) & (df['Sex'] == 'male')].head())

print("\n=== Количество n-местных кабин ===")
cabins = df.dropna(subset=['Cabin']).copy()
cabins['Cabin'] = cabins['Cabin'].apply(lambda x: str(x).split()[0])
cabin_counts = cabins['Cabin'].value_counts()
grouped = cabin_counts.value_counts().sort_index()
print(grouped)
```
