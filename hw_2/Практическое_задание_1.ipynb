```python
class Chocolate:
  def __init__(self, price: float, supl:str = "Без добавок"):
    self.price = price
    self.supl = supl
    if self.price <= 0:
      raise ValueError('Цена не может быть меньше нуля!')

  def show_my_chocolate(self):
    if self.supl == "Без добавок":
      print("Обычный шоколадный батончик")
    else:
      print(f"Шоколадный батончик, добавка: {self.supl}")

  def __str__(self):
    return f"Шоколадный батончик, {self.price} рублей"

```

```python
mars = Chocolate(56.34, "Карамель")
```

```python
print(mars)
```

```python
mars.show_my_chocolate()
```

```python
snikers = Chocolate(456)
```

```python
snikers.show_my_chocolate()
```

```python
print(snikers)
```
